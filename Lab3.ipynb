```python
def is_even(n):
    return n % 2 == 0
def count_evens(numbers):
    count = 0
    for number in numbers:
        print(number, "is even:", is_even(number))
        if is_even(number):
            count += 1
    return count
numbers = [2, 5, 8, 11, 14]

result = count_evens(numbers)
print("Number of even numbers:", result)
```

```python
def min_max(numbers):
    smallest = numbers[0]
    largest = numbers[0]
    for i in range(1, len(numbers)):
        if numbers[i] < smallest:
            smallest = numbers[i]
        elif numbers[i] > largest:
            largest = numbers[i]
    return (smallest, largest)

numbers = [3, 4, 6, 9, 2, 31, ]

answer = min_max(numbers)

print("Tuple:", answer)
print("Small:", answer[0])
print("Large:", answer[1])
```

```python
def letter_count(word):
    counts = {}

    for letter in word.lower():
        if letter.isalpha():
            counts[letter] = counts.get(letter, 0) + 1

    return counts

user_word = input("Enter a word: ")
result = letter_count(user_word)

print(" frequencies of letter:", result)
```

```python
def total_cost(products):
    total = 0
    for item_name, price in products:
        total += price

    return total

def mostas_expensive(products):
    if not products:
        return None

    expensive_item, highest_price = products[0]

    for item_name, price in products[1:]:
        if price > highest_price:
            expensive_item = item_name
            highest_price = price

    return expensive_item
products = [
    ("brush", 20), ("sunblock", 30),("eraser", 10)
]

print("Total cost:", total_cost(products))
print("Most expensive item:", mostas_expensive(products))
```

```python
def class_average(grades):
    total = 0

    for grade in grades.values():
        total += grade

    return total / len(grades)


def passing_students(grades, passing_mark=50):
    passed = []

    for name in grades:
        if grades[name] >= passing_mark:
            passed.append(name)

    return passed


grades = {}

while True:
    name = input("Enter name, or type done: ").strip()

    if name.lower() == "done":
        break

    grade = float(input("Enter grade: "))
    grades[name] = grade

print("Class average:", class_average(grades))
print("Passing students:", passing_students(grades))
```
